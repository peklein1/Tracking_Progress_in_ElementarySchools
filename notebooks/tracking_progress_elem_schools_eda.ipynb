```python
# Write a function that calculates percentages of missing values
def calculate_missing_values(df):
    """
    Calculate the number and percentage of missing values for each column in a dataframe.
    """
    
    missing_count = df.isna().sum()
    missing_percent = (missing_count / len(df)) * 100
    
    missing_summary = pd.DataFrame({
        "missing_count": missing_count,
        "missing_percent": missing_percent
    })
    
    return missing_summary.sort_values(
        "missing_percent",
        ascending=False
    )

missing_summary = calculate_missing_values(df)

missing_summary
```

## Build SQL Relational Tables

```python
# Load CSV files into pandas DataFrames (Replace file names and table names with actual file names and table names)
df_table_1 = pd.read_csv("../data/cleaned/file_1.csv")
df_table_2 = pd.read_csv("../data/cleaned/file_2.csv")
df_table_3 = pd.read_csv("../data/cleaned/file_3.csv")

# Connect to SQLite database
conn = sqlite3.connect("../database_name.db")

# Select columns for each DataFrame
df_table_1 = df_table_1[[
    'primary_key', 'column_1', 'column_2', 'column_3', 
    'column_4', 'column_5'
]]

df_table_2 = df_table_2[[
    'primary_key', 'foreign_key', 'column_1', 'column_2', 'column_3', 
    'column_4', 'column_5'
]]

df_table_3 = df_table_3[[
    'primary_key', 'foreign_key', 'column_1', 'column_2', 'column_3', 
    'column_4', 'column_5'
]]

# Write DataFrames to SQLite tables
df_table_1.to_sql("table_1", conn, index=False, if_exists='replace')
df_table_2.to_sql("table_2", conn, index=False, if_exists='replace')
df_table_3.to_sql("table_3", conn, index=False, if_exists='replace')

# Commit changes to the database
conn.commit()
```

## SQL Queries

### SQL Query 1: Basic SELECT

```python
query1 = """
SELECT *
FROM {TABLE_NAME}
LIMIT 10;
"""
result1 = pd.read_sql_query(query1, conn)
result1
```

### SQL Query 2: Filtering with WHERE

```python
query2 = """
SELECT *
FROM {TABLE_NAME}
WHERE 1 = 1
AND some_column = 'some_value'
AND numeric_column > 100
LIMIT 100;
"""
result2 = pd.read_sql_query(query2, conn)
result2
```

### SQL Query 3: GROUP BY and Aggregation

```python
query3 = """
SELECT
    some_category_column,
    COUNT(*) AS event_count,
    AVG(some_numeric_column) AS avg_value
FROM {TABLE_NAME}
GROUP BY some_category_column
ORDER BY event_count DESC;
"""
result3 = pd.read_sql_query(query3, conn)
result3
```

### SQL Query 4: CASE Expression

```python
query4 = """
SELECT
    some_numeric_column,
    CASE
        WHEN some_numeric_column < 10 THEN 'Low'
        WHEN some_numeric_column < 50 THEN 'Medium'
        ELSE 'High'
    END AS value_group
FROM {TABLE_NAME};
"""
result4 = pd.read_sql_query(query4, conn)
result4
```

### SQL Query 5: Common Table Expression

```python
query5 = """
WITH summary AS (
    SELECT
        some_category_column,
        COUNT(*) AS event_count,
        AVG(some_numeric_column) AS avg_value
    FROM {TABLE_NAME}
    GROUP BY some_category_column
)
SELECT *
FROM summary
ORDER BY avg_value DESC;
"""
result5 = pd.read_sql_query(query5, conn)
result5
```

### SQL Query 6: Window Function

```python
query6 = """
SELECT
    some_category_column,
    some_numeric_column,
    RANK() OVER (
        PARTITION BY some_category_column
        ORDER BY some_numeric_column DESC
    ) AS category_rank
FROM {TABLE_NAME};
"""
result6 = pd.read_sql_query(query6, conn)
result6
```

### SQL Query 7: Subquery

```python
query7 = """
SELECT *
FROM {TABLE_NAME}
WHERE some_numeric_column > (
    SELECT AVG(some_numeric_column)
    FROM {TABLE_NAME}
);
"""
result7 = pd.read_sql_query(query7, conn)
result7
```

### SQL Query 8: JOIN

```python
query8 = """
SELECT
    a.*,
    b.*
FROM table_a AS a
JOIN table_b AS b
    ON a.id = b.id
LIMIT 100;
"""
result8 = pd.read_sql_query(query8, conn)
result8
```
